```python
#INTEGRANTES: 

#SE DEFINEN FUNCIONES A OCUPAR:

def verificar(X): #Funcion de verificación para salas y cursos
    while True: #ayuda a pedir las salas y cursos hasta que se ingrese lo contrario
        if 1<=X<=100:#si ese X que puede ser sala o curso se encuentra entre 1 y 100 entonces, me devuelve el resultado
            return X
        else:
            X=int(input("Solo numeros entre 1 y 100: ")) # de lo contrario me pide que ingrese el numero denuevo hasta cumplir la condicion

#FUNCION VERIFCAR HORARIO
def verificarH(I,F):#funcion para verificación de salas y cursos, como entrada I que es hora inicial y F que es hora final
    while True:
        if 0<=I<F<=24: #debe cumplirse que aquella hora final debe ser mayor que 0 y mayor que hora inicial pero menor a 24hrs
            return I,F
        else:
            I=int(input("Hora de inicio mayor que 0hrs y menor que hora final: "))# De lo contrario pregunta por ambas horas denuevo hasta cunplir lo requerido
            F=int(input("Hora de final debe ser menor que 24hrs y mayor que hora de inicio: "))

#FUNCION VERIFICAR ID
def verificarID(D): #Funcion para verificar el ID, como debe estar compuesto de strings y sin espacio, esta verificación garantiza eso.
    while True: 
        if " " in D: # si se encuentra el espacio que es definido como " "
            D=input("Ingrese ID (sin espacios): ") #entonces me pide ingresar el string denuevo
        else:
            D=str(D)
            D=D.upper()
             # de lo contrarior hacemos que el return se transforme en mayuscula para tener todo en un mismo formato
            return D
            break 

def curso():#se define la funcion del curso ya que al ser recursivo y tener varias veces que ingresar datos, ayuda a simplificar el ingreso de curso
    ID_curso =input("Ingrese el ID del curso: ")#se ingresa el id de curso con string, haciendo que este ademas este en mayusculas
    ID_curso=verificarID(ID_curso) #lo que hace luego es verificar que ID calce con los que queremos, en el caso de que no cumpla requisito, se pueda asignar un nuevo ID guardandose ahora como el nuevo ID
    students = int(input("Ingrese el número de estudiantes: ")) #se ingresa numero de estudiantes
    schedule_inicio = int(input("Ingrese el bloque de inicio: "))#se ingresa horario inicial para el curso
    schedule_final = int(input("Ingrese el bloque de termino: "))#se ingresa horario final para el curso
    schedule_inicio,schedule_final=verificarH(schedule_inicio,schedule_final)#aqui verificamos las horas iniciales y finales, en caso de no calzar, las pide denuevo y las asigna como los nuevos horarios al igual como lo que sucedia con el ID
    requisitos=None #esto es sumamente importante porque cuando se ingresaban los datos y se ponia requisitos 0, asignaba como que no existia tal valor, por lo que fue mejor asignarlo antes de que cambie su valor como NONE para evitar problemas
    lista_requisitos=[] #lista para guardar los requisitos del curso
    lista_horas_curso=[] #lista para guardar las horas del curso
    lista_horas_curso.append(schedule_inicio) #se encarga de guardar la hora inicio
    lista_horas_curso.append(schedule_final)# se encarga de guardar hora final
    numero_requisitos= int(input("Ingrese numero requisitos")) #pide el numero de requisitos del curso
    if numero_requisitos==0:#en el caso de que el numero de requisitos sea 0, Se enlista dentro de requisitos como None, era conveniente dejarlo asi,porque cuando daba 0 con el None de las caracteristicas no aparecia, ahora se queria que apareciera para que cuente como un vacio dentro de la lista de requisitos 
        lista_requisitos.append('NONE')
    else:
        for i in range(numero_requisitos): #de lo contrario si no tiene problemas para el rango de 0 a el numero de requisitos se ejecutara numero de requisitos veces
            requisitos = input("Ingrese los requisitos del curso: ")
            requisitos=str(requisitos)
            requisitos=requisitos.upper()
            lista_requisitos.append(requisitos)#se encarga de incorporar los requisitos a la lista de requisitos
    return ID_curso,students,lista_horas_curso,numero_requisitos,requisitos,lista_requisitos #los datos que quiero que me devuelva para utilizar luego

def sala():#se define la funcion de las propiedades de las sala 
    ID_sala = (str(input("Ingrese el ID de la sala: "))).upper() #se define el id de la sala como un string y metodo .upper para que se haga mayusculas y no haya problema de comparaciones despues
    capacidad = int(input("Ingrese la capacidad de la sala: "))#se define la capacidad de estudiantes que soporta la sala
    caracteristicas=None #nuevamente se define none por el problema de que al asignar 0 caracteristicas, retornaba que no existia el valor, por tanto ahora cuando se ingresa 0, retornana un valor nulo
    lista_caracteristicas=[]#se crea la lista de caracteristicas
    lista_horas=list(range(25))#crea n listas por cada sala que se cree.
    numero_caracteristicas=int(input('Ingrese numero de caracteristicas'))#se ingresa numero de caracteristicas
    if numero_caracteristicas==0:#en caso de ser 0 el numero de caracteristicas
        lista_caracteristicas.append('NONE')  #como el valor nulo no aparece dentro de la lista, era mejor incorporarlo fisicamente para que se rellene la lista de caracteristicas, al igual que con la lista de requisitos  
    else:#de lo contrario va a ciclar una i cantidad de veces en el rango del numero de caracteristicas
         for i in range(numero_caracteristicas):        
             caracteristicas =input("Ingrese las características de la sala: ")
             caracteristicas=str(caracteristicas) #conversion a string
             caracteristicas=caracteristicas.upper()
             lista_caracteristicas.append(caracteristicas)#luego las caracteristicas se van adhiriendo a la lista de caracteristicas
    return ID_sala,capacidad,numero_caracteristicas,caracteristicas,lista_caracteristicas,lista_horas #me retorna los valores que quiero ocupar para ocupar luego


#SE INTRODUCEN LOS DATOS DE CANTIDAD DE CURSOS Y SALAS A ITERAR:

numero_cursos = int(input("Ingrese el número de cursos a agregar: ")) #se inserta numero de cursos
numero_cursos=verificar(numero_cursos) # si al insertar numeros falla, al ingresar el nuevo numero quedara guardado como el que se utilizara
numero_salas = int(input("Ingrese el número de salas a agregar: "))#ingresar numero de salas
numero_salas=verificar(numero_salas)#verifica si el numero de salas cumple con los requisitos de la verificacion, en caso de no serlo, el nuevo valor asignado sera el que se guardara si es que cumple los requisitos tambien.

#CREACION DE LISTA PARA GUARDAR CURSOS Y SALAS CREADAS:

Datos_C_1=[] #aqui se guardaran las listas con todo el curso en cada indice, por ejemplo el indice 0 contendra toda la info que quiero del curso1
Datos_S_1=[]#aqui se guardaran las listas con todo el curso en cada sala, por ejemplo


#LOGICA PARA PODER RELLENAR LOS DATOS RESPECTO A LA CANTIDAD DE CURSOS Y SALAS CORRESPONDIENTEMENTE ASIGNADAS

for i in range(numero_cursos):#por la cantidad de grupos ingresados, se itera esa cantidad de veces la funcion, que define los cursos.
    ID_curso,students,lista_horas_curso,numero_requisitos,requisitos,lista_requisitos=curso() #descomprimimos la funcio n
    Datos_C= [ID_curso]+[lista_requisitos]+[students]+[lista_horas_curso]#se compilan los datos del curso inicialmente creado en una lista,
    Datos_C_1.append(Datos_C)#luego esa lista creada se ingresa en otra lista que contendra todos los cursos.
    
for i in range(numero_salas):#se itera por la cantidad de numeros de solas
    ID_sala,capacidad,numero_caracteristicas,caracteristicas,lista_caracteristicas,lista_horas=sala()#descomprimimos esta función también
    Datos_S=[ID_sala]+[lista_caracteristicas]+[capacidad]+[lista_horas]#luego las sala creada se contendra en una lista
    Datos_S_1.append(Datos_S)#esa sala creada contenida en una lista ahora se integrara a otra lista que contendra el conjunto de listas creadas.
#ejemplo_: lista=[], se crea una sala con datos x,y,z, eso se encerrara en una lista sala1[x,y,z], y eso se metera dentro de otra lista [sala1,sala2...]


#CREACION DE LISTA PARA CURSOS ASIGNADOS Y NO ASIGNADOS

Cursos_asignados=[] #lista para filtrar a los cursos asignados
Cursos_no_asignados=[] # lista para filtrar a aquellos cursos no asignados
S=0 #Cursos asignados exitosamente
F=0 #Cursos no asignados exitosamente

#LOGICA DE ASIGNACION DATOS PARA SALAS Y LUEGO LOS CURSOS, ENTONCES COMPARA CADA CURSO CON UNA SALA. Y LUEGO DE COMPARAR CURSOS CON 1 SALA PASA A LA SIGUENTE.

for i in range(numero_cursos): #se iterara (numero de cursos) veces los datos de adentro para poder ir rotando y seleccionando la informacion correspondiente
    students=int(Datos_C_1[i][2])#se accede a los datos de la sublista, en este caso accede al curso numero i creado, y accede al objeto en el indice 2 de la lista, que es la cantidad de estudiantes
    Inicio=int(Datos_C_1[i][3][0])#se accede a los datos de la sublista, en este caso accede al curso numero i creado, seleccionando objeto en indice 3 y como el indice tiene sublista, accede a la posicion 0 de esa sublista. siendo la hora inicial
    Fin=int(Datos_C_1[i][3][1])#se accede a los datos de la sublista igual que en el dato anterior, pero ahora accede al indice 1 de la lista del horario contenida en el curso, que posee la hora final.
    req_curso=Datos_C_1[i][1]#se accede a los datos de la sublista del curso i, y los requisitos se encuentran en el indice 1. Para luego comparar
    n=len(Datos_C_1[i][1]) # numero de objetos que posee cada curso_i
    curso_asignado=False#todos los cursos de antemano son asignados como que no califican. Haciendo semejanza a un guardia de una disco, no sabe quien va a entrar pero hasta que nadie se presente y cumpla todos los requisitos 
    for j in range(numero_salas):#para el numero de salas se itera lo siguente, que sigue una logica similar a lo de arriba.
        lista_horas=Datos_S_1[j][3]#se va checkeando las horas disponibles de las n salas creadas. porque luego se iran removiendo las horas de rango digitadas.
        capacidad=int(Datos_S_1[j][2])#se define la capacidad como el digito, que se encuentra respecto a la sala j, en el indice 2
        carac_sala=Datos_S_1[j][1]# se definen las caracteristicas de las salas como los datos dentro de la lista en el indice 1


#LOGICA DE FILTRACIÓN DE CURSOS SE ENCARGA DE FILTRAR LOS CURSOS QUE SEAN ASIGNADOS
        
        if Inicio in lista_horas and Fin in lista_horas and students<=capacidad: #luego de definir las variables a las que se quiere acceder y además definir la logica para asignar las salas. La esencia del loop anterior muestra que los a Cursos_i se iran viendo si calzan o no dentro de la sala_j. como curso 1, curso 2, curso 3, verificar si calzan con sala1, luego los mismo cursos si calza en sala 2 y asi sucesivamente para i cursos y j salas.
            switch=True #al ingresar a esta condicion se "prende" el primer switch y pasa al siguente filtro.
            if req_curso != ['NONE']: #al acceder a los requisitos del curso i, si esta lista de requisitos posee elementos, entonces pasara al siguente filtro
                    for req in req_curso: #iterando sobre los elementos del curso
                        if req not in carac_sala:# si es que estos no se encuentrar en las caracteristicas de la sala, entonces  
                            switch=False# el switch se vuelve falso y ya no clasifica como apto para ESA sala. y pasara al siguente curso.
                            break#pasara al siguente curso
#LUEGO DE QUE SEAN ASIGNADOS
            if switch==True:#luego como el switch es verdadero si cumplio lo anterior
                Cursos_asignados.append(Datos_C_1[i][0])# Por tanto se asigna el curso a una lista de cursos asignados
                Cursos_asignados.append(Datos_S_1[j][0])#junto a este se asigna la sala al lado de el curso, quedando ordenado en curso -> sala, SIEMPRE.
                for k in range(Inicio,Fin):# luego como ha sido asignado exitosamente se removeran las horas disponibles de la sala asignadas en el rango del horario.
                    lista_horas.remove(k)
                curso_asignado=True#finalmmente el curso adquiere condicion de verdad.
                break#se rompe el loop y pasa a lo siguente.
    if curso_asignado==False:#si el curso reprueba todos los filtros, o reprueba aunque sea uno, es considerado instantaneamente no apto para asignación.
        Cursos_no_asignados.append(Datos_C_1[i][0])#se asignara el curso a la lista de no asignados.


#logica final para el conteo de cursos
S=int(len(Cursos_asignados)/2) #se define la longitud de la lista los cursos asignados, como estos vienen tambien con una sala asignada, es necesario dividir por 2 para obtener solo las clases.
F=int(len(Cursos_no_asignados)) #se define la longitud de la lista de cursos no asignados. Ambos son transformados a int porque luego tiene problema el for loop para leer numeros que no sean enteros.
print(S) #muestra el numero de cursos asignados
for b in range(S): #por cada curso asignado, mostrar el curso junto con su sala asignada
    print(Cursos_asignados[b*2],Cursos_asignados[(b*2)+1]) #va saltando la lista de cursos asignados de 2 en 2, es decir: CURSO A,LISTA A lee primero luego CURSO B ,LISTA B. De una lista general CURSO A,LISTA A, CURSO B, LISTA B
print(F) #muestra el numero de cursos no asignados
for a in range(F): #por cada curso no asignado, los imprimme a cada uno
    print(Cursos_no_asignados[a])




#INTEGRANTES:
#Hector Andres Bascur Irribarra
#Pierre Benjamin Bertin Reyes
#Cristobal Benjamin Ferreira Ojeda
#Nicolas Jesus Neira
```
